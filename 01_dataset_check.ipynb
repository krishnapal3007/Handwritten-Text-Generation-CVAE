```python
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model

print("TensorFlow version:", tf.__version__)

# Load preprocessed dataset
X_train = np.load("models/X_train.npy")
X_val = np.load("models/X_val.npy")
y_train = np.load("models/y_train.npy")
y_val = np.load("models/y_val.npy")
classes = np.load("models/classes.npy", allow_pickle=True)

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("y_train:", y_train.shape)
print("y_val:", y_val.shape)
print("Classes:", classes)
print("Number of classes:", len(classes))
```

```python

IMG_SIZE = 64
LATENT_DIM = 32
NUM_CLASSES = len(classes)

# Add channel dimension
X_train = X_train.astype("float32")
X_val = X_val.astype("float32")

if X_train.max() > 1:
    X_train = X_train / 255.0
    X_val = X_val / 255.0

X_train = X_train[..., np.newaxis]
X_val = X_val[..., np.newaxis]

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("Number of classes:", NUM_CLASSES)
```

```python
# =============================
# ENCODER
# =============================

image_input = layers.Input(
    shape=(IMG_SIZE, IMG_SIZE, 1),
    name="image_input"
)

label_input = layers.Input(
    shape=(),
    dtype=tf.int32,
    name="label_input"
)

# Image encoder
x = layers.Conv2D(32, 3, strides=2, padding="same", activation="relu")(image_input)
x = layers.Conv2D(64, 3, strides=2, padding="same", activation="relu")(x)
x = layers.Conv2D(128, 3, strides=2, padding="same", activation="relu")(x)

x = layers.Flatten()(x)

# Label embedding
label_embedding = layers.Embedding(
    NUM_CLASSES,
    32
)(label_input)

label_embedding = layers.Flatten()(label_embedding)

# Combine image + character label
x = layers.Concatenate()([x, label_embedding])

x = layers.Dense(256, activation="relu")(x)

z_mean = layers.Dense(LATENT_DIM, name="z_mean")(x)
z_log_var = layers.Dense(LATENT_DIM, name="z_log_var")(x)


# Reparameterization trick
def sampling(args):
    z_mean, z_log_var = args

    epsilon = tf.random.normal(
        shape=tf.shape(z_mean)
    )

    return z_mean + tf.exp(0.5 * z_log_var) * epsilon


z = layers.Lambda(
    sampling,
    name="z"
)([z_mean, z_log_var])


encoder = Model(
    [image_input, label_input],
    [z_mean, z_log_var, z],
    name="encoder"
)

encoder.summary()
```

```python
# =============================
# DECODER
# =============================

latent_input = layers.Input(
    shape=(LATENT_DIM,),
    name="latent_input"
)

decoder_label = layers.Input(
    shape=(),
    dtype=tf.int32,
    name="decoder_label"
)

# Character embedding
label_embedding = layers.Embedding(
    NUM_CLASSES,
    32
)(decoder_label)

label_embedding = layers.Flatten()(label_embedding)

# Combine latent vector + character
x = layers.Concatenate()([
    latent_input,
    label_embedding
])

x = layers.Dense(8 * 8 * 128, activation="relu")(x)

x = layers.Reshape((8, 8, 128))(x)

x = layers.Conv2DTranspose(
    128, 3, strides=2,
    padding="same",
    activation="relu"
)(x)

x = layers.Conv2DTranspose(
    64, 3, strides=2,
    padding="same",
    activation="relu"
)(x)

x = layers.Conv2DTranspose(
    32, 3, strides=2,
    padding="same",
    activation="relu"
)(x)

output_image = layers.Conv2D(
    1,
    3,
    padding="same",
    activation="sigmoid"
)(x)

decoder = Model(
    [latent_input, decoder_label],
    output_image,
    name="decoder"
)

decoder.summary()
```

```python
# ==========================================
# IMPROVED CVAE
# ==========================================

class CVAE(Model):

    def __init__(self, encoder, decoder, **kwargs):
        super().__init__(**kwargs)

        self.encoder = encoder
        self.decoder = decoder

        self.total_loss_tracker = tf.keras.metrics.Mean(
            name="loss"
        )

        self.reconstruction_loss_tracker = tf.keras.metrics.Mean(
            name="reconstruction_loss"
        )

        self.kl_loss_tracker = tf.keras.metrics.Mean(
            name="kl_loss"
        )

    @property
    def metrics(self):
        return [
            self.total_loss_tracker,
            self.reconstruction_loss_tracker,
            self.kl_loss_tracker
        ]

    def calculate_loss(
        self,
        images,
        labels,
        training=False
    ):

        z_mean, z_log_var, z = self.encoder(
            [images, labels],
            training=training
        )

        reconstructed = self.decoder(
            [z, labels],
            training=training
        )

        # ==================================
        # Reconstruction loss
        # ==================================

        bce = tf.keras.backend.binary_crossentropy(
            images,
            reconstructed
        )

        # Give more importance to handwriting pixels
        pixel_weight = 1.0 + 8.0 * images

        weighted_bce = bce * pixel_weight

        reconstruction_loss = tf.reduce_mean(
            tf.reduce_sum(
                weighted_bce,
                axis=(1, 2, 3)
            )
        )

        # ==================================
        # KL loss
        # ==================================

        kl_loss = -0.5 * tf.reduce_mean(
            tf.reduce_sum(
                1 + z_log_var
                - tf.square(z_mean)
                - tf.exp(z_log_var),
                axis=1
            )
        )

        # Very small KL weight
        beta = 0.0001

        total_loss = (
            reconstruction_loss
            + beta * kl_loss
        )

        return (
            total_loss,
            reconstruction_loss,
            kl_loss
        )

    def train_step(self, data):

        images, labels = data

        with tf.GradientTape() as tape:

            total_loss, reconstruction_loss, kl_loss = \
                self.calculate_loss(
                    images,
                    labels,
                    training=True
                )

        gradients = tape.gradient(
            total_loss,
            self.trainable_weights
        )

        self.optimizer.apply_gradients(
            zip(
                gradients,
                self.trainable_weights
            )
        )

        self.total_loss_tracker.update_state(
            total_loss
        )

        self.reconstruction_loss_tracker.update_state(
            reconstruction_loss
        )

        self.kl_loss_tracker.update_state(
            kl_loss
        )

        return {
            "loss":
                self.total_loss_tracker.result(),

            "reconstruction_loss":
                self.reconstruction_loss_tracker.result(),

            "kl_loss":
                self.kl_loss_tracker.result()
        }

    def test_step(self, data):

        images, labels = data

        total_loss, reconstruction_loss, kl_loss = \
            self.calculate_loss(
                images,
                labels,
                training=False
            )

        return {
            "loss": total_loss,
            "reconstruction_loss": reconstruction_loss,
            "kl_loss": kl_loss
        }

    def call(self, inputs):

        images, labels = inputs

        z_mean, z_log_var, z = self.encoder(
            [images, labels],
            training=False
        )

        return self.decoder(
            [z, labels],
            training=False
        )
```

```python
cvae = CVAE(
    encoder,
    decoder
)

cvae.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0003
    )
)

print("Improved CVAE ready!")
```

```python
# ==========================================
# HANDWRITTEN CHARACTER GENERATOR
# ==========================================

def generate_character(character, temperature=1.0):

    # Check character
    matches = np.where(classes == character)[0]

    if len(matches) == 0:
        raise ValueError(
            f"'{character}' dataset ki classes me nahi hai."
        )

    class_index = int(matches[0])

    # Your decoder expects latent dimension = 32
    latent_dim = 32

    # Random latent vector
    z = tf.random.normal(
        shape=(1, latent_dim)
    ) * temperature

    # Character label
    label = np.array(
        [class_index],
        dtype=np.int32
    )

    # Generate handwritten character
    generated = decoder.predict(
        [z, label],
        verbose=0
    )[0]

    # Display
    plt.figure(figsize=(4, 4))

    plt.imshow(
        generated.squeeze(),
        cmap="gray",
        vmin=0,
        vmax=1
    )

    plt.axis("off")
    plt.title(f"Generated: {character}")

    plt.show()

    return generated
```

```python
# ==========================================
# READABLE HANDWRITTEN TEXT GENERATOR
# ==========================================

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageFilter

def generate_readable_text(text, scale=1.0, spacing=4):

    text = str(text)

    images = []

    for char in text:

        # Space between words
        if char == " ":
            images.append(None)
            continue

        matches = np.where(classes == char)[0]

        if len(matches) == 0:
            print(f"Skipping unsupported character: {char}")
            continue

        class_index = int(matches[0])

        # Get real examples of this character
        indices = np.where(y_train == class_index)[0]

        if len(indices) == 0:
            continue

        # Random real handwriting sample
        idx = np.random.choice(indices)

        img = X_train[idx].squeeze()

        # Convert to uint8
        img = (img * 255).astype(np.uint8)

        # Resize
        pil_img = Image.fromarray(img)

        new_size = (
            int(64 * scale),
            int(64 * scale)
        )

        pil_img = pil_img.resize(
            new_size,
            Image.Resampling.LANCZOS
        )

        images.append(pil_img)

    # --------------------------------------
    # Calculate canvas width
    # --------------------------------------

    width = 0

    for img in images:

        if img is None:
            width += 35
        else:
            width += img.width + spacing

    height = 64

    canvas = Image.new(
        "L",
        (max(width, 100), height),
        0
    )

    # --------------------------------------
    # Paste characters
    # --------------------------------------

    x = 0

    for img in images:

        if img is None:
            x += 35
            continue

        canvas.paste(
            img,
            (x, 0)
        )

        x += img.width + spacing

    # --------------------------------------
    # Display
    # --------------------------------------

    plt.figure(
        figsize=(max(10, width / 40), 3)
    )

    plt.imshow(
        canvas,
        cmap="gray",
        vmin=0,
        vmax=255
    )

    plt.axis("off")
    plt.title(text)

    plt.show()

    return canvas
```
